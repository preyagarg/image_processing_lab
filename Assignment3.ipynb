# Assignment 3

1. Draw a border around the input image. Take width and border color from the user. Try for binary and grayscale images.
2. Find image complement for binary and grayscale images.


```python
import cv2
import numpy as np

img = cv2.imread('images/sample_image.png')
B = img[:, :, 0].astype(np.float64)
G = img[:, :, 1].astype(np.float64)
R = img[:, :, 2].astype(np.float64)

gray = ((R + G + B) / 3).astype(np.uint8)
mean_threshold = int(np.mean(gray))
binary = np.zeros_like(gray)
binary[gray >= mean_threshold] = 255
```

## 1. Border / Padding


```python
def add_border(image, width, value):
    h, w = image.shape
    result = np.full((h + 2*width, w + 2*width), value, dtype=np.uint8)
    result[width:width+h, width:width+w] = image
    return result

border_width = int(input('Enter border width in pixels: '))
gray_color = int(input('Enter grayscale border value (0-255): '))
binary_color = int(input('Enter binary border value (0 or 255): '))
gray_border = add_border(gray, border_width, gray_color)
binary_border = add_border(binary, border_width, binary_color)
cv2.imwrite('outputs/assignment3/grayscale_border.jpg', gray_border)
cv2.imwrite('outputs/assignment3/binary_border.jpg', binary_border)
```

## 2. Image Complement


```python
gray_complement = 255 - gray
binary_complement = 255 - binary
cv2.imwrite('outputs/assignment3/grayscale_complement.jpg', gray_complement)
cv2.imwrite('outputs/assignment3/binary_complement.jpg', binary_complement)
```
