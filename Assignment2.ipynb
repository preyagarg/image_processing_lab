# Assignment 2

Convert RGB image to grayscale.

- Case 1: Mean of three planes.
- Case 2: User-defined R, G, B weights whose sum is 1.


```python
import cv2
import numpy as np

img = cv2.imread('images/sample_image.jpg')
B = img[:, :, 0].astype(np.float64)
G = img[:, :, 1].astype(np.float64)
R = img[:, :, 2].astype(np.float64)
print('Input image loaded:', img.shape)
```

## Case 1: Mean


```python
gray_mean = ((R + G + B) / 3).astype(np.uint8)
cv2.imwrite('outputs/assignment2/grayscale_mean.jpg', gray_mean)
```

## Case 2: User-defined weights


```python
wr = float(input('Enter weight for R: '))
wg = float(input('Enter weight for G: '))
wb = float(input('Enter weight for B: '))
if 0 <= wr <= 1 and 0 <= wg <= 1 and 0 <= wb <= 1 and abs(wr + wg + wb - 1) < 1e-6:
    gray_weighted = wr * R + wg * G + wb * B
    gray_weighted = np.clip(gray_weighted, 0, 255).astype(np.uint8)
    cv2.imwrite('outputs/assignment2/grayscale_weighted.jpg', gray_weighted)
    print('Weights accepted:', wr, wg, wb)
else:
    print('Invalid weights')
```
