# Assignment 1

Write a program to convert a grayscale image to a binary image.

- Case 1: Mean intensity as threshold.
- Case 2: Threshold value from the user.


```python
import cv2
import numpy as np

img = cv2.imread('images/sample_image.jpg')
B = img[:, :, 0].astype(np.float64)
G = img[:, :, 1].astype(np.float64)
R = img[:, :, 2].astype(np.float64)

gray = ((R + G + B) / 3).astype(np.uint8)
print('Grayscale image created')
```

## Case 1: Mean intensity threshold


```python
mean_threshold = int(np.mean(gray))
binary_mean = np.zeros_like(gray)
binary_mean[gray >= mean_threshold] = 255
cv2.imwrite('outputs/assignment1/binary_mean_threshold.jpg', binary_mean)
print('Mean threshold:', mean_threshold)
```

## Case 2: User threshold


```python
threshold = int(input('Enter threshold value (0-255): '))
binary_user = np.zeros_like(gray)
binary_user[gray >= threshold] = 255
cv2.imwrite('outputs/assignment1/binary_user_threshold.jpg', binary_user)
print('Threshold used:', threshold)
```
